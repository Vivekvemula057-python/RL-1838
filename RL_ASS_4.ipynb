```python

import math

# Part A: Machine Initialization
machines = ['A', 'B', 'C']
N_play = 10
c = 2

# Part B: Initialize N, R, UCB and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

# Part C: UCB Selection
for play in range(1, N_play + 1):

    print("\n========== PLAY", play, "==========")

    # First play every machine once
    unplayed = [m for m in machines if N[m] == 0]

    if unplayed:
        selected = unplayed[0]
    else:
        for m in machines:
            UCB[m] = Q[m] + c * math.sqrt(math.log(play) / N[m])

        print("\nUCB Values:")
        for m in machines:
            print(m, "=", round(UCB[m], 2))

        selected = max(UCB, key=UCB.get)

    print("\nSelected Machine:", selected)

    # Part D: Reward Collection
    reward = float(input("Enter reward: "))

    # Part E: Update N, R and Q
    N[selected] += 1
    R[selected].append(reward)

    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    total_reward += reward

    print("Updated N:", N)
    print("Updated Q:", {m: round(Q[m], 2) for m in machines})


# Calculate final UCB values
for m in machines:
    if N[m] > 0:
        UCB[m] = Q[m] + c * math.sqrt(math.log(N_play) / N[m])
    else:
        UCB[m] = float('inf')


# Part F: Display Final Results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tCount\tRewards\t\tUCB\t\tQ Value")

for m in machines:
    print(m, "\t", N[m], "\t", R[m],
          "\t", round(UCB[m], 2), "\t", round(Q[m], 2))

print("\nTotal Reward Received:", total_reward)

print("Best Machine:", max(Q, key=Q.get))

```
