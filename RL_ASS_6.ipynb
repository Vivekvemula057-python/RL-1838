```python
import gymnasium as gym

# Part A: Environment Initialization
environments = {
    "CliffWalking": "CliffWalking-v1",
    "Taxi": "Taxi-v4"
}

episodes = 10

# Run both environments
for name, env_id in environments.items():

    print("\n================================")
    print("Environment:", name)
    print("==============================")

    # Create environment
    env = gym.make(env_id)

    total_all_episodes = 0

    # Repeat for multiple episodes
    for episode in range(1, episodes + 1):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        steps = 0

        print("\nEpisode:", episode)
        print("Initial State:", state)
        print("Available Actions:", env.action_space)

        # Random Agent
        while True:

            # Select random action
            action = env.action_space.sample()

            # Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            print("State:", state,
                  "Action:", action,
                  "Reward:", reward,
                  "Next State:", next_state)

            # Update state and reward
            total_reward += reward
            steps += 1
            state = next_state

            # Check termination
            if terminated or truncated:
                break

        # Calculate average reward
        average_reward = total_reward / steps if steps > 0 else 0

        total_all_episodes += total_reward

        print("\nEpisode Completed!")
        print("Total Steps:", steps)
        print("Total Reward:", total_reward)
        print("Average Reward:", round(average_reward, 2))

    # Final results
    print("\n========== FINAL RESULTS ==========")
    print("Environment:", name)
    print("Number of Episodes:", episodes)
    print("Total Reward of All Episodes:", total_all_episodes)
    print("Average Episode Reward:",
          round(total_all_episodes / episodes, 2))

    env.close()

print("\nBoth Environments Completed!")
```
