```python
import random

machines = ['A', 'B', 'C']

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0, 'B': 0, 'C': 0}

M = 3
N_play = 10
epsilon = 0.2

total_reward = 0

print("===== Modified Epsilon-Greedy Multi-Armed Bandit =====")

for play in range(1, N_play + 1):

    r = random.random()

    highest_Q = max(Q.values())
    best_machines = [m for m in machines if Q[m] == highest_Q]

    if r < epsilon:
        exploration_machines = [
            m for m in machines if m not in best_machines
        ]
        selected_machine = random.choice(exploration_machines)
        decision = "Exploration"
    else:
        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Play with machine", selected_machine)

    reward = float(input("Enter reward: "))

    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    total_reward += reward

    print("Updated N:", N)
    print("Updated R:", R)
    print("Updated Q:", Q)

print("\n========== FINAL RESULTS ==========")

print("\nParticipation Count (N):")
for machine in machines:
    print(machine, ":", N[machine])

print("\nTotal Rewards (R):")
for machine in machines:
    print(machine, ":", R[machine])

print("\nEstimated Q-values:")
for machine in machines:
    print(machine, ":", round(Q[machine], 2))

print("\nTotal Reward Received:", total_reward)
```
