```python
import random
machines = ['A', 'B', 'C']
M = 3
N_play = 10
epsilon = 0.2
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

print("======================================")
print("   ε-GREEDY MULTI-ARMED BANDIT")
print("======================================")
print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)
print()

for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        selected_machine = random.choice(machines)
        action = "Exploration"
    else:
        max_q = max(Q.values())

        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        action = "Exploitation"

    print(f"Play {play}")
    print(f"Random number = {r:.4f}")
    print(f"Mode = {action}")
    print(f"Play with machine {selected_machine}")

    reward = float(input("Enter reward: "))

    N[selected_machine] += 1

    R[selected_machine].append(reward)

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    total_reward += reward

    print(f"Updated Q({selected_machine}) = "
          f"{Q[selected_machine]:.2f}")
    print("-" * 40)


print("\n======================================")
print("           FINAL RESULTS")
print("======================================")

print("\nN TABLE – Participation Count")
print("--------------------------------------")
for machine in machines:
    print(f"Machine {machine}: {N[machine]} times")

print("\nR TABLE – Rewards")
print("--------------------------------------")
for machine in machines:
    print(f"Machine {machine}: {R[machine]}")

print("\nQ TABLE – Estimated Q-Values")
print("--------------------------------------")
for machine in machines:
    print(f"Machine {machine}: {Q[machine]:.2f}")

print("\nTotal Reward =", total_reward)

best_machine = max(Q, key=Q.get)

print("Best Machine =", best_machine)
print("Highest Q-value =", round(Q[best_machine], 2))
```
