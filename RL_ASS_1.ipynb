```python

n = int(input("Enter number of machines: "))

machines = []

for i in range(n):
    name = input(f"\nEnter name of Machine {i + 1}: ")

    rewards = list(map(float, input(
        f"Enter rewards for Machine {name} separated by spaces: "
    ).split()))

    machines.append((name, rewards))

q_values = {}

for name, rewards in machines:
    q = sum(rewards) / len(rewards)
    q_values[name] = q

print("\n========== RESULTS ==========")

for name, rewards in machines:
    print(f"\nMachine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q_values[name]:.2f}")

selected_machine = max(q_values, key=q_values.get)

print("\n==============================")
print(f"Greedy Selected Machine: {selected_machine}")
print("==============================")
```
