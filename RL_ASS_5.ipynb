```python

import gymnasium as gym

# List of environments
environments = {
    "CartPole": "CartPole-v1",
    "MountainCar": "MountainCar-v0",
    "FrozenLake": "FrozenLake-v1"
}

# Run Random Agent in all environments
for name, env_id in environments.items():

    print("\n==============================")
    print("Environment:", name)
    print("==============================")

    # Part 1: Create environment
    env = gym.make(env_id)

    # Part 2: Reset environment
    state, info = env.reset()

    print("Initial State:", state)
    print("Available Actions:", env.action_space)

    total_reward = 0
    steps = 0

    # Part 3: Random Agent
    while True:

        # Part 4: Select random action
        action = env.action_space.sample()

        # Part 5: Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        # Part 6: Display interaction
        print("\nStep:", steps + 1)
        print("Current State:", state)
        print("Selected Action:", action)
        print("Reward:", reward)
        print("Next State:", next_state)

        # Update reward and state
        total_reward += reward
        steps += 1
        state = next_state

        # Part 7: Check episode termination
        if terminated or truncated:
            print("\nEpisode Finished!")
            break

    # Part 8: Calculate results
    average_reward = total_reward / steps if steps > 0 else 0

    print("\n========== FINAL RESULTS ==========")
    print("Environment:", name)
    print("Total Steps:", steps)
    print("Total Reward:", total_reward)
    print("Average Reward:", round(average_reward, 2))

    # Close environment
    env.close()

print("\nAll Environments Completed!")
```
