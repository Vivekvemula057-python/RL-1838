```python

# Assignment: Policy Iteration in Deterministic Grid World

# Part A: Environment Initialization

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.0001

# Rewards
normal_reward = -1
wall_reward = -1
goal_reward = 100

# Initial State Values
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

# Initial Policy
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

# Part B: Transition Function
def transition(state, action):

    if state == 'G':
        return 'G', 0

    # Transitions from S0
    if state == 'S0':

        if action == 'Down':
            return 'S1', normal_reward

        elif action == 'Right':
            return 'S0', wall_reward

        else:
            return 'S0', normal_reward

    # Transitions from S1
    elif state == 'S1':

        if action == 'Up':
            return 'S0', normal_reward

        elif action == 'Right':
            return 'G', goal_reward

        else:
            return 'S1', normal_reward


# Part C: Policy Evaluation
def policy_evaluation():

    while True:

        max_change = 0

        for state in ['S0', 'S1']:

            old_value = V[state]

            action = policy[state]

            next_state, reward = transition(state, action)

            V[state] = reward + gamma * V[next_state]

            change = abs(old_value - V[state])

            max_change = max(max_change, change)

        if max_change < theta:
            break


# Part D: Calculate Action Values
def calculate_q_values():

    Q = {}

    for state in ['S0', 'S1']:

        Q[state] = {}

        for action in actions:

            next_state, reward = transition(state, action)

            Q[state][action] = reward + gamma * V[next_state]

    return Q


# Part E: Policy Improvement
def policy_improvement():

    stable = True

    Q = calculate_q_values()

    for state in ['S0', 'S1']:

        old_action = policy[state]

        best_action = max(Q[state], key=Q[state].get)

        policy[state] = best_action

        if old_action != best_action:
            stable = False

    return stable


# Part F: Policy Iteration
iteration = 1

while True:

    print("\n========== ITERATION", iteration, "==========")

    # Evaluate current policy
    policy_evaluation()

    print("\nState Values:")

    for state in states:
        print("V(" + state + ") =", round(V[state], 2))

    # Calculate action values
    Q = calculate_q_values()

    print("\nAction Values:")

    for state in ['S0', 'S1']:
        for action in actions:
            print("Q(" + state + "," + action + ") =",
                  round(Q[state][action], 2))

    # Improve policy
    stable = policy_improvement()

    print("\nUpdated Policy:", policy)

    if stable:
        print("\nPolicy is Stable!")
        break

    iteration += 1


# Part G: Final Results
print("\n========== FINAL OPTIMAL RESULTS ==========")

print("\nOptimal State Values:")

for state in states:
    print("V(" + state + ") =", round(V[state], 2))

print("\nOptimal Policy:")

for state in ['S0', 'S1']:
    print(state, "->", policy[state])

print("\nOptimal Action Values:")

Q = calculate_q_values()

for state in ['S0', 'S1']:
    for action in actions:
        print("Q(" + state + "," + action + ") =",
              round(Q[state][action], 2))

print("\nOptimal Path:")
print("S0 -> S1 -> G")

print("\nTotal Iterations:", iteration)

print("\nProgram Completed Successfully!")
```
