```python

print("===== ASSIGNMENT 1 =====")

n = int(input("Enter number of time steps: "))

rewards = []

for i in range(n):
    r = float(input(f"Enter reward for time step {i+1}: "))
    rewards.append(r)

gamma = float(input("Enter discount factor (gamma): "))

# Calculate Undiscounted Return
undiscounted = sum(rewards)

# Calculate Discounted Return
discounted = 0

for i in range(n):
    discounted += (gamma ** i) * rewards[i]

# Display Results
print("\n===== RETURN RESULTS =====")
print("Entered Rewards:", rewards)
print("Undiscounted Return:", undiscounted)
print("Discounted Return:", round(discounted, 2))



print("\n===== ASSIGNMENT 2: GRID WORLD MDP =====")

# Part A: State Space
states = ['S0', 'S1', 'G', 'X', 'S3', 'S4']

# Part B: Action Space
actions = ['up', 'down', 'left', 'right']

# Part C: Discount Factor
gamma = 0.9

# Part D: Reward Function
def reward_function(state, action, next_state):

    if next_state == 'G':
        return 10

    elif next_state == 'X':
        return -10

    else:
        return -1


# Part E: Transition Function
transitions = {

    'S0': {
        'up': 'S0',
        'down': 'S0',
        'left': 'S0',
        'right': 'S1'
    },

    'S1': {
        'up': 'S1',
        'down': 'S3',
        'left': 'S0',
        'right': 'G'
    },

    'S3': {
        'up': 'S1',
        'down': 'S3',
        'left': 'S3',
        'right': 'S4'
    },

    'S4': {
        'up': 'G',
        'down': 'S4',
        'left': 'S3',
        'right': 'S4'
    },

    'G': {
        'up': 'G',
        'down': 'G',
        'left': 'G',
        'right': 'G'
    },

    'X': {
        'up': 'X',
        'down': 'X',
        'left': 'X',
        'right': 'X'
    }
}


# Part F: Estimate Next State
def get_next_state(current_state, action):

    if current_state not in states:
        return "Invalid State"

    if action not in actions:
        return "Invalid Action"

    return transitions[current_state][action]


# Display MDP Details
print("\nState Space:", states)
print("Action Space:", actions)
print("Discount Factor:", gamma)

# Display Transition Table
print("\n===== TRANSITION TABLE =====")

for state in states:
    for action in actions:
        print(state, "+", action, "->",
              get_next_state(state, action))


# Get User Input
print("\n===== GRID WORLD SIMULATION =====")

current_state = input("Enter current state (S0, S1, S3, S4, G, X): ").upper()

action = input("Enter action (up/down/left/right): ").lower()

# Calculate Next State
next_state = get_next_state(current_state, action)

if next_state == "Invalid State" or next_state == "Invalid Action":

    print(next_state)

else:

    reward = reward_function(current_state, action, next_state)

    print("\n===== MDP RESULT =====")
    print("Current State:", current_state)
    print("Selected Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)
    print("Discount Factor:", gamma)

    if next_state == 'G':
        print("Goal Reached!")

    elif next_state == 'X':
        print("Obstacle Reached!")

print("\n===== PROGRAM COMPLETED =====")
```
