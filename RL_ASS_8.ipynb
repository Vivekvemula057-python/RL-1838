```python

# Assignment: Calculate State Value and Action Value
# 2 x 3 Grid World using Policy and Discounted Return

# Part A: Environment Initialization

states = ['S0', 'S1', 'S3', 'S4', 'G']
actions = ['up', 'down', 'left', 'right']

gamma = 0.9

# Grid layout
grid = [
    ['S0', 'S1', 'G'],
    ['X', 'S3', 'S4']
]

# Part B: Define Policy
policy = {
    'S0': 'right',
    'S1': 'down',
    'S3': 'right',
    'S4': 'up'
}

# Part C: Transition Function
def transition(state, action):

    row, col = None, None

    # Find current position
    for i in range(2):
        for j in range(3):
            if grid[i][j] == state:
                row, col = i, j

    # Movement directions
    if action == 'up':
        row -= 1
    elif action == 'down':
        row += 1
    elif action == 'left':
        col -= 1
    elif action == 'right':
        col += 1

    # Check grid boundaries
    if row < 0 or row >= 2 or col < 0 or col >= 3:
        return state, -1, False

    next_state = grid[row][col]

    # Check obstacle
    if next_state == 'X':
        return state, -5, False

    # Check goal
    if next_state == 'G':
        return 'G', 10, True

    # Normal movement
    return next_state, -1, False


# Part D: State Value Function
def state_value_function(start_state):

    state = start_state
    total_return = 0
    discount = 1

    while state != 'G':

        action = policy[state]

        next_state, reward, done = transition(state, action)

        total_return += discount * reward

        discount *= gamma

        state = next_state

        if done:
            break

    return total_return


# Part E: Action Value Function
def action_value_function(state, action):

    total_return = 0
    discount = 1

    # Execute given action first
    next_state, reward, done = transition(state, action)

    total_return += reward
    discount *= gamma

    state = next_state

    # Follow policy for remaining steps
    while state != 'G' and not done:

        action = policy[state]

        next_state, reward, done = transition(state, action)

        total_return += discount * reward

        discount *= gamma

        state = next_state

    return total_return


# Part F: Display State Values
print("===== GRID WORLD MDP =====")

print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Policy:", policy)

print("\n===== STATE VALUES =====")

for state in states:
    if state != 'G':
        value = state_value_function(state)
        print("V(" + state + ") =", round(value, 3))

# Part G: Display Action Value
print("\n===== ACTION VALUE =====")

value = action_value_function('S0', 'up')

print("State: S0")
print("Action: up")
print("Q(S0, up) =", round(value, 3))

print("\n===== PROGRAM COMPLETED =====")
```
