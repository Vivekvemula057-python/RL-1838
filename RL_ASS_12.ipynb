```python

# Assignment: Value Iteration in Deterministic Grid World

# Part A: Environment Initialization

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.0001

# Rewards
normal_reward = -1
wall_reward = -1
goal_reward = 100

# Initialize State Values
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

# Part B: Transition Function
def transition(state, action):

    if state == 'G':
        return 'G', 0

    # Transitions from S0
    if state == 'S0':

        if action == 'Down':
            return 'S1', normal_reward

        else:
            return 'S0', wall_reward

    # Transitions from S1
    elif state == 'S1':

        if action == 'Up':
            return 'S0', normal_reward

        elif action == 'Right':
            return 'G', goal_reward

        else:
            return 'S1', normal_reward


# Part C: Calculate Action Values
def calculate_q(state, action):

    next_state, reward = transition(state, action)

    q_value = reward + gamma * V[next_state]

    return q_value


# Part D: Value Iteration Algorithm
iteration = 0

while True:

    delta = 0
    new_values = V.copy()

    for state in ['S0', 'S1']:

        old_value = V[state]

        # Calculate Q values for all actions
        q_values = []

        for action in actions:
            q = calculate_q(state, action)
            q_values.append(q)

        # Update state value using maximum Q value
        new_values[state] = max(q_values)

        delta = max(delta, abs(old_value - new_values[state]))

    V = new_values
    iteration += 1

    print("\nIteration:", iteration)
    print("V(S0) =", round(V['S0'], 4))
    print("V(S1) =", round(V['S1'], 4))

    # Check convergence
    if delta < theta:
        break


# Part E: Calculate Final Action Values and Optimal Policy
Q = {}
policy = {}

for state in ['S0', 'S1']:

    Q[state] = {}

    for action in actions:

        Q[state][action] = calculate_q(state, action)

    # Select action with maximum Q value
    best_action = max(Q[state], key=Q[state].get)

    policy[state] = best_action


# Part F: Display Final Results

print("\n========== FINAL RESULTS ==========")

print("\nOptimal State Values:")

for state in states:
    print("V(" + state + ") =", round(V[state], 2))


print("\n========== ACTION VALUE TABLE ==========")

for state in ['S0', 'S1']:

    print("\nState:", state)

    for action in actions:
        print("Q(" + state + "," + action + ") =",
              round(Q[state][action], 2))


print("\n========== OPTIMAL POLICY ==========")

for state in ['S0', 'S1']:
    print(state, "->", policy[state])


print("\nOptimal Path:")
print("S0 -> S1 -> G")

print("\nTotal Iterations:", iteration)

print("\nValue Iteration Completed Successfully!")
```
